```python
import pandas as pd
from sklearn.metrics import mean_absolute_error

df = pd.read_csv("../data/historical_trains.csv")
df.shape
```

```python
# Naive baseline: assume future delay = current delay (no change)
baseline_predictions = df['current_delay_min']
actual_values = df['future_delay_min']

baseline_mae = mean_absolute_error(actual_values, baseline_predictions)
print(f"Baseline MAE: {baseline_mae:.2f} minutes")
```

```python
# Alternative baseline: average future delay per segment (ignores individual conditions)
segment_avg_delay = df.groupby(['origin_station', 'destination_station'])['future_delay_min'].transform('mean')

avg_baseline_mae = mean_absolute_error(actual_values, segment_avg_delay)
print(f"Segment-Average Baseline MAE: {avg_baseline_mae:.2f} minutes")
```

```python
import json

baseline_results = {
    "naive_baseline_mae": round(baseline_mae, 2),
    "segment_avg_baseline_mae": round(avg_baseline_mae, 2)
}

with open("../data/baseline_results.json", "w") as f:
    json.dump(baseline_results, f, indent=2)

print("Baseline results saved!")
```
